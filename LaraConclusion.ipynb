## 8. Conclusion

The simulation of the time evolution of the XXZ Heisenberg Hamiltonian was done using Trotterisation and McLachan, concluding that Trotterisation was a much better method to be using than a variational one for many body qubit systems in our simulations due to the structure of the ansatze in common variational methods.

Noise was investigated alongside the simulation of qubits for our trotter (for testing) and the qiskit trotter (for comparison to IBM data). These showed us how errors on quantum data appear in quantum computers and what this does to the data. Including investigations into the bit and phase flips that change quantum infmation in a qubit, and decoherent noise where all errors act with equal probability. 

Simulating a larger number of qubits in many body simulations still remains challange when aiming for accuracy of data while maximising data points. Our IBM data focuses just on the accuracy of points of collectable data of the evolved qubit states over a short time period of $5 eV^{-1}$. These results validated our noise models since they were far noisier than or classical and trotterisation methods of evolution, proving that quantum computer hardware noise is a massive contributing factor to errors in reeults from simulations. 

Incorporating reserch on reducing this noise, like quantum error correction, or looking into more advanced noise models for specific conditions of qubit temperatures, states and interactions could be a way to improve quantum simulations in future resreach.
