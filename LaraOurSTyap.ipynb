## 3. Our built Suzuki-Trotter function

Simulating the time evolution of quantum many body systems on quantum computers has proved to be far more useful for larger number of bodies and timescales than a classical system. We have used the Suzuki-Trotter method to simulate this motion in Qiskit using own implementation of a quantum S-T decomposition circuit to fully understand how the model works. 

The Suzuki-Trotter decomposition approximates the time evolution operator by splitting the Hamiltonian into individually exponentiable terms[@trotter3]:

$$U(t) = e^{-iHt} \approx \left(\prod_k e^{-i h_k \Delta t}\right)^n + \mathcal{O}(\Delta t^2)$$

where $H = \sum_k h_k$ and $\Delta t = t/n$. Each $e^{-ih_k \Delta t}$ is a simple rotation gate directly implementable on quantum hardware.

Trotterisation maps naturally onto quantum circuits with fixed, shallow depth per step. For comparison to IBM hardware the circuit depth must remain fixed and we therefor fix the Trotter circuit structure (controlled by `reps`), and vary $\Delta t$ to sample different times. This keeps the circuit depth constant and matches how the IBM jobs are actually submitted. 

The error in the approximation decreases as $\Delta t \to 0$ (more Trotter steps), but on hardware, smaller $\Delta t$ means more repetitions and deeper circuits. While higher order $\Delta t$  means bettter scaling, they require more gates per Trotter step, which on noisy hardware is not effective for collecting usable data. This makes the Suzuki-Trotter method a good model for hardware in quantum computation for small intervals of time.
